### Line Detection

```python
from google.colab import files
from IPython.display import Image
import cv2
import numpy as np

uploaded = files.upload()

for fn in uploaded.keys():
  print('User uploaded file "{name}" with length {length} bytes'.format(
      name=fn, length=len(uploaded[fn])))
  display(Image(fn))

  # Read the uploaded image using OpenCV
  image = cv2.imread(fn)

  gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

  # Deteksi tepi menggunakan Canny
  edges = cv2.Canny(gray, 50, 150, apertureSize=3)

  # Deteksi garis menggunakan Hough Transform
  # Menyesuaikan ambang batas (threshold) untuk mendapatkan hasil yang lebih baik
  lines = cv2.HoughLines(edges, 1, np.pi / 180, 80) # Ubah threshold menjadi 200 untuk menyaring lebih banyak garis

  # Buat salinan gambar asli untuk menggambar garis
  image_with_lines = image.copy()

  if lines is not None:
    # Tentukan panjang garis yang sesuai dengan dimensi gambar
    img_height, img_width = image.shape[:2]
    line_length = max(img_height, img_width)

    for rho, theta in lines[:, 0]:
      a = np.cos(theta)
      b = np.sin(theta)
      x0 = a * rho
      y0 = b * rho
      x1 = int(x0 + line_length * (-b))
      y1 = int(y0 + line_length * (a))
      x2 = int(x0 - line_length * (-b))
      y2 = int(y0 - line_length * (a))
      # (0, 0, 255) berarti warna Merah dalam format BGR
      cv2.line(image_with_lines, (x1, y1), (x2, y2), (0, 0, 255), 2)

  # Simpan hasil deteksi garis pada salinan gambar
  cv2.imwrite('line_detected_image.jpg', image_with_lines)
  display(Image('line_detected_image.jpg'))
```

### Bungkus Jadi API dengan FastAPI

[Documentation](https://fastapi.tiangolo.com/)

Kelima operasi image processing di atas (Cropping, Grayscale, Channel Split, Convolution, Line Detection) bisa dibungkus jadi REST API pakai FastAPI. Gambar diproses langsung di memori (`BytesIO`), tanpa perlu disimpan ke disk — hasilnya dikembalikan dalam format **base64**, karena endpoint API tidak bisa langsung mengembalikan file gambar mentah.

#### Setup & Helper Function

```python
from fastapi import FastAPI, UploadFile, File
from PIL import Image as PILImage, ImageOps
import io
import base64
import cv2
import numpy as np

app = FastAPI()

# Helper function: konversi PIL image ke base64
def pil_image_to_base64(image: PILImage.Image):
    buffered = io.BytesIO()
    image.save(buffered, format="JPEG")
    return base64.b64encode(buffered.getvalue()).decode("utf-8")

# Helper function: konversi OpenCV image ke base64
def cv2_image_to_base64(image):
    _, buffer = cv2.imencode('.jpg', image)
    return base64.b64encode(buffer).decode("utf-8")
```

#### Endpoint 1 — Image Cropping (`/crop/`)

```python
@app.post("/crop/")
async def crop_image(file: UploadFile = File(...), x: int = 0,
                      y: int = 0, width: int = 100, height: int = 100):
    image = PILImage.open(io.BytesIO(await file.read()))
    cropped_image = image.crop((x, y, x + width, y + height))
    return {"image_base64": pil_image_to_base64(cropped_image)}
```

#### Endpoint 2 — Grayscale (`/grayscale/`)

```python
@app.post("/grayscale/")
async def grayscale_image(file: UploadFile = File(...)):
    image = PILImage.open(io.BytesIO(await file.read()))
    grayscale_image = ImageOps.grayscale(image)
    return {"image_base64": pil_image_to_base64(grayscale_image)}
```

#### Endpoint 3 — Channel Split (`/channel_split/`)

```python
@app.post("/channel_split/")
async def channel_split(file: UploadFile = File(...), channel: str = "red"):
    image = cv2.imdecode(np.frombuffer(await file.read(), np.uint8),
                          cv2.IMREAD_COLOR)
    (blue, green, red) = cv2.split(image)
    if channel == "red":
        return {"image_base64": cv2_image_to_base64(red)}
    elif channel == "green":
        return {"image_base64": cv2_image_to_base64(green)}
    else:
        return {"image_base64": cv2_image_to_base64(blue)}
```

#### Endpoint 4 — Convolution / Blur (`/convolution/`)

```python
@app.post("/convolution/")
async def convolution(file: UploadFile = File(...)):
    image = cv2.imdecode(np.frombuffer(await file.read(),
                          np.uint8), cv2.IMREAD_COLOR)
    kernel = np.ones((5, 5), np.float32) / 25
    convolved_image = cv2.filter2D(image, -1, kernel)
    return {"image_base64": cv2_image_to_base64(convolved_image)}
```

#### Endpoint 5 — Line Detection (`/line_detection/`)

```python
@app.post("/line_detection/")
async def line_detection(file: UploadFile = File(...)):
    image = cv2.imdecode(np.frombuffer(await file.read(),
                          np.uint8), cv2.IMREAD_COLOR)
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    edges = cv2.Canny(gray, 50, 150, apertureSize=3)
    lines = cv2.HoughLines(edges, 1, np.pi / 180, 200)

    if lines is not None:
        for rho, theta in lines[:, 0]:
            a = np.cos(theta)
            b = np.sin(theta)
            x0 = a * rho
            y0 = b * rho
            x1 = int(x0 + 1000 * (-b))
            y1 = int(y0 + 1000 * (a))
            x2 = int(x0 - 1000 * (-b))
            y2 = int(y0 - 1000 * (a))
            cv2.line(image, (x1, y1), (x2, y2), (0, 0, 255), 2)

    return {"image_base64": cv2_image_to_base64(image)}
```

#### Endpoint 6 — Proses Gambar Lokal (misalnya, Grayscale) `/process_local_grayscale/{filename}`

Endpoint ini akan memproses gambar yang sudah tersimpan di direktori Colab (`/content/`). Anda hanya perlu memberikan nama file gambar sebagai bagian dari URL.

```python
@app.get("/process_local_grayscale/{filename}")
async def process_local_grayscale(filename: str):
    file_path = f"/content/{filename}"
    # Cek apakah file ada
    if not os.path.exists(file_path):
        raise HTTPException(status_code=404, detail="Image not found")

    # Baca gambar menggunakan OpenCV
    image = cv2.imread(file_path)
    if image is None:
        raise HTTPException(status_code=400, detail="Could not read image file. Is it a valid image?")

    # Konversi ke grayscale
    grayscale_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Kembalikan gambar grayscale dalam format base64
    return {"image_base64": cv2_image_to_base64(grayscale_image)}
```

#### Bonus: Endpoint Versi "Gambar Langsung" (bukan Base64)

Endpoint-endpoint sebelumnya balikin JSON berisi base64 — cocok kalau API-nya dipanggil dari aplikasi lain (web/mobile), tapi kurang enak dilihat langsung di browser/Swagger UI karena cuma nongol sebagai teks panjang.

Versi di bawah ini fungsinya sama persis, cuma **response-nya langsung file gambar** (`media_type="image/jpeg"`) — jadi kalau dites di Swagger UI, hasilnya langsung kelihatan sebagai gambar, dan kalau dibuka lewat browser/`<img src="...">`, langsung ke-render juga.

#### (Opsional) Menjalankan Server FastAPI Sungguhan dari Colab

Kalau mau expose API-nya jadi URL publik dari Colab (misalnya buat dites dari Postman atau aplikasi lain di luar notebook), pakai `nest_asyncio` + `uvicorn` + `pyngrok`. Colab butuh `nest_asyncio` karena kernelnya sudah menjalankan satu event loop, sementara `uvicorn.run()` butuh event loop-nya sendiri.

```python
!pip install nest-asyncio pyngrok -q
```

```python
import nest_asyncio
import uvicorn
from pyngrok import ngrok
import asyncio

nest_asyncio.apply()

# Ganti dengan authtoken asli dari:
# https://dashboard.ngrok.com/get-started/your-authtoken
NGROK_AUTHTOKEN = "PASTE_AUTHTOKEN_ASLI_DI_SINI"
ngrok.set_auth_token(NGROK_AUTHTOKEN)

# --- LAPIS 1: matikan instance server & task SEBELUMNYA (kalau ada) dulu ---
# Ini yang mencegah dua uvicorn rebutan port 8000 di percobaan berikutnya.
_old_server = globals().get('_uvicorn_server')
if _old_server is not None:
    _old_server.should_exit = True

_old_task = globals().get('_server_task')
if _old_task is not None and not _old_task.done():
    _old_task.cancel()

ngrok.kill()  # tutup semua tunnel ngrok lokal yang masih nyangkut

_uvicorn_server = None

async def run_fastapi_server():
    global _uvicorn_server
    try:
        # ngrok.connect() adalah fungsi SINKRON biasa, jangan pakai await di sini
        public_url = ngrok.connect(8000)
        print("=" * 60)
        print("API bisa diakses publik di:", public_url)
        print("Coba buka:", str(public_url).split('"')[1] + "/docs")
        print("=" * 60)

        config = uvicorn.Config(app, host="0.0.0.0", port=8000, loop="asyncio")
        _uvicorn_server = uvicorn.Server(config)
        await _uvicorn_server.serve()
    except (Exception, SystemExit) as e:
        # LAPIS 2: SystemExit HARUS ikut ditangkap di sini.
        # uvicorn versi baru memanggil sys.exit() kalau gagal bind port (mis. port
        # masih dipakai instance lama), dan SystemExit BUKAN turunan Exception --
        # kalau cuma "except Exception", dia lolos dan bisa bikin kernel Colab
        # crash & auto-restart, persis seperti yang barusan kejadian.
        print("Gagal menjalankan server:", type(e).__name__, "-", e)

_server_task = asyncio.create_task(run_fastapi_server())

# Kasih waktu buat task-nya jalan & nge-print URL sebelum cell ini "selesai"
await asyncio.sleep(3)
```
